```python
import pandas as pd

# Q1: Build Personalized Knowledge Base

roll_number = input("Enter your roll number: ")
last_two = roll_number[-2:]

categories = ["billing", "account", "general"]

fixed_entries = [
    {
        "question": "what is the annual fee",
        "answer": "The annual fee is Rs 500.",
        "keywords": "fee cost price charge",
        "category": "billing"
    },
    {
        "question": "how to reset password",
        "answer": "Go to Settings > Reset Password.",
        "keywords": "password reset login",
        "category": "account"
    },
    {
        "question": "what are your working hours",
        "answer": "We are open 9 AM to 5 PM.",
        "keywords": "hours timing open time",
        "category": "general"
    },
    {
        "question": "how can i pay the fee",
        "answer": "You can pay via UPI, card, or net banking.",
        "keywords": "pay payment upi fee",
        "category": "billing"
    }
]

personalized_entries = []

for digit in last_two:
    d = int(digit)
    category = categories[d % 3]

    if category == "billing":
        question = "how can i check my payment status"
        answer = "You can check your payment status in the billing section."
        keywords = "payment status billing transaction"

    elif category == "account":
        question = "how do i update my registered mobile number"
        answer = "Go to Account Settings and update your registered mobile number."
        keywords = "mobile number update account"

    else:
        question = "where can i get general support"
        answer = "You can contact the support team for general assistance."
        keywords = "support help assistance"

    personalized_entries.append({
        "question": question,
        "answer": answer,
        "keywords": keywords,
        "category": category
    })
all_entries = fixed_entries + personalized_entries

df = pd.DataFrame(all_entries)

print("\nFinal 6-row DataFrame:")
print(df)
```

```python
#q2
def score_query(query, df):
    query_words = set(query.lower().split())

    results = []

    for index, row in df.iterrows():
        keywords = set(row["keywords"].lower().split())

        score = len(query_words.intersection(keywords))

        if score > 0:
            results.append((score, row["question"], row["answer"]))

    results.sort(reverse=True)

    return results


query = input("\nEnter your query: ")

results = score_query(query, df)

print("\nMatching entries:")

for score, question, answer in results:
    print("Score:", score)
    print("Question:", question)
    print("Answer:", answer)
    print()
```

```python
#q3
def same_category(category_name, df):
    return df[df["category"] == category_name]["question"]

category_name = personalized_entries[0]["category"]

print("\nQuestions in category:", category_name)
print(same_category(category_name, df))
```

```python
#q4
print("\nKnowledge Base:")
print(df)
index = 0

new_keyword = input("\nEnter a new keyword: ")


df.loc[index, "keywords"] = df.loc[index, "keywords"] + " " + new_keyword

filename = roll_number + "_faq_data.csv"
df.to_csv(filename, index=False)

print("\nUpdated DataFrame:")
print(df)

print("\nFile saved as:", filename)
```

```python
#q5
print("\nNumber of FAQ entries per category:")

category_count = df.groupby("category").size()

print(category_count)
```

```python
#q6
def best_match(query, df):

    query_words = set(query.lower().split())

    scores = []

    for index, row in df.iterrows():

        keywords = set(row["keywords"].lower().split())

        score = len(query_words.intersection(keywords))

        scores.append(score)

    max_score = max(scores)

    if max_score == 0:
        print("No matching entry found.")
        return

    print("\nHighest Score:", max_score)
    print("Matching entries:")

    for index, row in df.iterrows():

        keywords = set(row["keywords"].lower().split())

        score = len(query_words.intersection(keywords))

        if score == max_score:
            print("\nQuestion:", row["question"])
            print("Answer:", row["answer"])
            print("Category:", row["category"])
```

```python
print("\n--- TIE QUERY ---")
best_match("fee", df)
```

```python
print("\n--- NON-TIE QUERY ---")
best_match("password", df)
```
