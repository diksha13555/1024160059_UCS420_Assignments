```python
#Q6 Create employees dataset
emp_data = {
    "Employee_ID": [101,102,103,104,105],
    "Name": ["Alice","Bob","Charlie","Diana","Edward"],
    "Department": ["HR","IT","IT","Marketing","Sales"],
    "Age": [29,34,41,28,38],
    "Salary": [50000,70000,65000,55000,60000],
    "Years_of_Experience": [4,8,10,3,12],
    "Joining_Date": ["2020-03-15","2017-07-19","2013-06-01","2021-02-10","2010-11-25"],
    "Gender": ["Female","Male","Male","Female","Male"],
    "Bonus": [5000,7000,6000,4500,5000],
    "Rating": [4.5,4.0,3.8,4.7,3.5]
}

emp_df = pd.DataFrame(emp_data)
emp_df.to_csv("employees.csv", index=False)
emp_df
```

```python
# a) Shape of the DataFrame
print("Shape (rows, columns):", emp_df.shape)
```

```python
# b) Summary of DataFrame - dtypes and non-null counts
emp_df.info()
```

```python
# c) Descriptive statistics
emp_df.describe()
```

```python
# d) First 5 rows and last 3 rows
print("First 5 rows:")
print(emp_df.head(5))
print("\nLast 3 rows:")
print(emp_df.tail(3))
```

```python
# e) Statistics
avg_salary = emp_df["Salary"].mean()
total_bonus = emp_df["Bonus"].sum()
youngest_age = emp_df["Age"].min()
highest_rating = emp_df["Rating"].max()

print("i.   Average Salary:", avg_salary)
print("ii.  Total Bonus Paid:", total_bonus)
print("iii. Youngest Employee's Age:", youngest_age)
print("iv.  Highest Rating:", highest_rating)
```

```python
# f) Sort by Salary descending
emp_df_sorted = emp_df.sort_values(by="Salary", ascending=False)
emp_df_sorted
```

```python
# g) Categorize employees by performance rating
def categorize(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

emp_df["Performance_Category"] = emp_df["Rating"].apply(categorize)
emp_df
```

```python
# h) Identify missing values
emp_df.isnull().sum()
```

```python
# i) Rename Employee_ID to ID
emp_df = emp_df.rename(columns={"Employee_ID": "ID"})
emp_df
```

```python
# j) Filter employees
# i. More than 5 years of experience
exp_filter = emp_df[emp_df["Years_of_Experience"] > 5]
print("Employees with more than 5 years of experience:")
print(exp_filter)

# ii. Belong to IT department
it_filter = emp_df[emp_df["Department"] == "IT"]
print("\nEmployees in IT department:")
print(it_filter)
```

```python
# k) Add Tax column - 10% of Salary
emp_df["Tax"] = emp_df["Salary"] * 0.10
emp_df
```

```python
# l) Save modified DataFrame to a new CSV file
emp_df.to_csv("employees_modified.csv", index=False)
print("Saved successfully as employees_modified.csv")
```
