```python
#Q1
roll_number = "1024160059"
digits = [int(d) for d in roll_number]
L = [d * 10 for d in digits]
print("L:", L)
L.append(99)
print("After append(99):", L)
L.insert(3, 77)
print("After insert(3, 77):", L)
L.remove(0)
print("After remove(0):", L)
popped = L.pop()
print(f"After pop() (removed {popped}):", L)
L.sort()
print("Ascending:", L)
L.sort(reverse=True)
print("Descending:", L)
print("First three:", L[:3])
print("Last three:", L[-3:])
avg = sum(L) / len(L)
above_avg = [x for x in L if x > avg]
print("Above average:", above_avg)
```

```python
#Q2
L_original = [d * 10 for d in digits]
scores = tuple(L_original[:8])
print("Scores:", scores)
highest = max(scores)
print(f"Highest: {highest} at index {scores.index(highest)}")
lowest = min(scores)
print(f"Lowest: {lowest}, appears {scores.count(lowest)} times")
reversed_scores = list(reversed(scores))
print("Reversed:", reversed_scores)
user_score = int(input("Enter a score to search: "))
print(f"Found at index {scores.index(user_score)}" if user_score in scores else "Not present")
try:
    scores[0] = 100
except TypeError as e:
    print("Error:", e)
first, second, *remaining = scores
print("First:", first, "Second:", second, "Remaining:", remaining)
```

```python
#Q3
import random
random.seed(1024160059)

random_list = [random.randint(100, 900) for _ in range(100)]  # i

odd_count = sum(1 for x in random_list if x % 2 != 0)
print("Odd numbers:", odd_count)  # ii

even_count = sum(1 for x in random_list if x % 2 == 0)
print("Even numbers:", even_count)  # iii

def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True

prime_list = [x for x in random_list if is_prime(x)]
print("Prime count:", len(prime_list))
print("Primes:", prime_list)  # iv

from collections import Counter
counts = Counter(random_list)
most_common_num, freq = counts.most_common(1)[0]
print(f"Most frequent: {most_common_num}, occurs {freq} times")  # v
```

```python
#Q4
first_eight_digits = digits[:8]
A = {d * 7 for d in first_eight_digits}
B = {d * 9 for d in first_eight_digits}
print("Set A:", A)
print("Set B:", B)
print("Union:", A | B)
print("Intersection:", A & B)
a_minus_b = A.difference(B)
b_minus_a = B.difference(A)
print("A - B:", a_minus_b)
print("B - A:", b_minus_a)
print("Symmetric difference:", A.symmetric_difference(B))
print("A subset of B:", A.issubset(B))
print("B superset of A:", B.issuperset(A))
X = int(input("Enter a value to remove from A: "))
A.discard(X)
print("A after discard:", A)
```
