```python
# Q1
import numpy as np
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
#a
corrected_temperature = temperature + 2
print(corrected_temperature)
#b
fahrenheit = (9/5) * corrected_temperature + 32
print(fahrenheit)
#c
greater_than_32 = corrected_temperature[corrected_temperature > 32]
print(greater_than_32)
#d
count = np.sum(corrected_temperature > 32)
print(count)
#e
#Vectorization performs operations on an entire NumPy array at once instead of explicitly processing each element with a for loop. This makes numerical operations shorter, faster, and generally more efficient.
#Boolean indexing creates a Boolean condition for every element and directly selects the values satisfying that condition. Therefore, operations such as finding all temperatures above 32°C can be performed without manually checking every reading.
```

```python
#q2
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])
#a
total = np.sum(steps)
print(total)
#b
mean = np.mean(steps)
print(mean)
#c
maximum = np.max(steps)
minimum = np.min(steps)

print("Maximum:", maximum)
print("Minimum:", minimum)
#d
daily_total = np.sum(steps, axis=0)
print(daily_total)
#e
user_total = np.sum(steps, axis=1)
print(user_total)
#f
position = np.unravel_index(np.argmax(steps), steps.shape)
print(position)
```

```python
#q3
import numpy as np

# a.
original = np.array([1, 2, 3, 4, 5, 6])
print("Original:", original)

# b.
subset = original[1:5]
print("Subset:", subset)

# c.
subset[0] = 999

print("After modifying subset:")
print("Original:", original)
print("Subset:", subset)

# d.
copied = original[1:5].copy()
copied[0] = 500

print("\nAfter modifying copied array:")
print("Original:", original)
print("Copied:", copied)

# e.
matrix = np.arange(1, 13).reshape(3, 4)
print("\n3 x 4 Matrix:")
print(matrix)

# f

# First row
print("\nFirst row:")
print(matrix[0])

# Last row
print("\nLast row:")
print(matrix[-1])

# Second column
print("\nSecond column:")
print(matrix[:, 1])

# Rows 1-2 and columns 2-3
print("\nRows 1-2 and columns 2-3:")
print(matrix[0:2, 1:3])

# g.
flattened = matrix.flatten()
raveled = matrix.ravel()

print("\nFlatten:")
print(flattened)

print("\nRavel:")
print(raveled)

# h.
raveled[0] = 100

print("\nAfter modifying ravel:")
print("Ravel:", raveled)
print("Original matrix:")
print(matrix)

# i.
flattened[1] = 200

print("\nAfter modifying flatten:")
print("Flatten:", flattened)
print("Original matrix:")
print(matrix)

# j.
print("\nArray properties:")
print("Shape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype)
```

```python
#q4
import numpy as np

X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])
#a
print("Shape:", X.shape)
print("Dimensions:", X.ndim)
#b
print(X.T)
#c
XtX = X.T @ X
print(XtX)
#d
XtX_inv = np.linalg.inv(XtX)
print(XtX_inv)
#e
beta = np.linalg.inv(X.T @ X) @ X.T @ y

print("Coefficients:")
print(beta)
#g
new_user = np.array([5, 40, 7])
prediction = new_user @ beta
print("Predicted assistance score:", prediction)
```
