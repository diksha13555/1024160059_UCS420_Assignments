```python
#5.2
n = int(input("Enter a number: "))
total = 0
for i in range(1, n + 1):
    if i % 7 == 0 and i % 9 == 0:
        total += i
print(f"Sum of numbers divisible by both 7 and 9 up to {n}: {total}")
```

```python
#5.3
n = int(input("Enter a number: "))
prime_sum = 0
for num in range(2, n + 1):
    is_prime = True
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        prime_sum += num
print(f"Sum of prime numbers up to {n}: {prime_sum}")
```

```python
#6.1
def sum_odd_numbers(n):
    total = 0
    for i in range(1, n + 1, 2):
        total += i
    return total

n = int(input("Enter a number: "))
print(f"Sum of odd numbers up to {n}: {sum_odd_numbers(n)}")
```

```python
#6.2
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True

def sum_primes(n):
    total = 0
    for i in range(2, n + 1):
        if is_prime(i):
            total += i
    return total

n = int(input("Enter a number: "))
print(f"Sum of prime numbers up to {n}: {sum_primes(n)}")
```
