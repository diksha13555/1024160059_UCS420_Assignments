```python
#q2bi
import numpy as np
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
values, counts = np.unique(x, return_counts=True)
max_count = counts.max()
most_frequent = values[counts == max_count]
print("Most frequent value:", most_frequent)
for value in most_frequent:
    print("Indices:", np.where(x == value)[0])
```

```python
#Q2bii
import numpy as np
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
values, counts = np.unique(y, return_counts=True)
max_count = counts.max()
most_frequent = values[counts == max_count]
print("Most frequent value:", most_frequent)
for value in most_frequent:
    print("Indices:", np.where(y == value)[0])
```

```python
#q4
import numpy as np

Diksha = np.linspace(10, 100, 25)

print("Array:")
print(Diksha)

print("Dimensions:", Diksha.ndim)
print("Shape:", Diksha.shape)
print("Total elements:", Diksha.size)
print("Data type:", Diksha.dtype)
print("Total bytes:", Diksha.nbytes)
transpose_array = Diksha.reshape(1, 25).T

print("Transpose using reshape():")
print(transpose_array)

print("Transpose using T attribute:")
print(Diksha.T)
```

```python
#q5
import numpy as np

ucs420_Diksha = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original Array:")
print(ucs420_Diksha)
print("Mean:", np.mean(ucs420_Diksha))
print("Median:", np.median(ucs420_Diksha))
print("Maximum:", np.max(ucs420_Diksha))
print("Minimum:", np.min(ucs420_Diksha))
print("Unique Elements:", np.unique(ucs420_Diksha))
reshaped_ucs420_Diksha = ucs420_Diksha.reshape(4, 3)

print("Reshaped Array (4 x 3):")
print(reshaped_ucs420_Diksha)
resized_ucs420_Diksha = np.resize(ucs420_Diksha, (2, 3))
print("Resized Array (2 x 3):")
print(resized_ucs420_Diksha)
```
